# Unit Conversions

```python
# Easy unit conversion and constants
from astropy import units as u
from astropy import constants as const
```

```python
c = const.c
c
```

```python
print(c)
```

```python
c_au_yr = c.to(u.AU/u.yr)
c_au_yr
```

```python
c_au_yr.value, c_au_yr.unit
```
